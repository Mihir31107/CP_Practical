```python
purity = float(input("Enter the purity of the product (as a percentage): "))

if not (0 <= purity <= 100):
    print(f"Rejected: Purity value {purity}% is invalid.")

else:
    pH = float(input("Enter the pH of the product: "))

    if not (0 <= pH <= 14):
        print(f"Rejected: pH value {pH} is invalid.")

    else:
        moisture = float(input("Enter the moisture content of the product (as a percentage): "))

        if not (0 <= moisture <= 100):
            print(f"Rejected: Moisture value {moisture}% is invalid.")

        else:
            # All values are valid
            print(f"Purity entered: {purity}%")
            print(f"pH entered: {pH}")
            print(f"Moisture entered: {moisture}%")

            if purity >= 98.0 and 6.5 <= pH <= 7.5 and moisture <= 1.0:
                classification = "Grade A"

            elif purity >= 95.0 and 6.0 <= pH <= 8.0 and moisture <= 2.0:
                classification = "Grade B"

            elif purity >= 90.0 and 5.0 <= pH <= 9.0 and moisture <= 4.0:
                classification = "Grade C"

            else:
                classification = "Rejected"

            print(f"The product is classified as: {classification}")
```

```python
#create a simple if elif else code to check marks and give 
#grade A,B,C,D,F based on the marks entered by the user.
marks = float(input('Enter your marks: '))

if marks >= 80:
    grade = 'A'
elif marks >= 60:
    grade = 'B'
elif marks >= 50:
    grade = 'C'
elif marks >= 40:
    grade = 'D'
else:
    grade = 'F'

print('Your grade is:', grade)
```

```python
purity = float(input("Enter the purity of the product (as a percentage): "))

if not (0 <= purity <= 100):
    print(f"Rejected: Purity value {purity}% is invalid.")

else:
    pH = float(input("Enter the pH of the product: "))

    if not (0 <= pH <= 14):
        print(f"Rejected: pH value {pH} is invalid.")

    else:
        moisture = float(input("Enter the moisture content of the product (as a percentage): "))

        if not (0 <= moisture <= 100):
            print(f"Rejected: Moisture value {moisture}% is invalid.")

        else:
            # Determine the grade
            if purity >= 98.0 and 6.5 <= pH <= 7.5 and moisture <= 1.0:
                classification = "A"

            elif purity >= 95.0 and 6.0 <= pH <= 8.0 and moisture <= 2.0:
                classification = "B"

            elif purity >= 90.0 and 5.0 <= pH <= 9.0 and moisture <= 4.0:
                classification = "C"

            else:
                classification = "Rejected"

            # Determine the status
            if classification == "Rejected":
                status = "REJECTED"
            else:
                status = "APPROVED FOR RELEASE"

            # Print the Chemical Quality Report
            print("------------------------------------------")
            print("          CHEMICAL QUALITY REPORT")
            print("------------------------------------------")
            print(f"Purity       : {purity:.2f} %")
            print(f"pH           : {pH:.2f}")
            print(f"Moisture     : {moisture:.2f} %")
            print(f"Final Grade  : {classification}")
            print(f"Status       : {status}")
            print("------------------------------------------")
```
